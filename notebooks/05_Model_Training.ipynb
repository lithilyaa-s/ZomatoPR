```python
# ===========================================
# Zomato Restaurant Rating Prediction
# Model Training
# ===========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

print("Libraries Imported Successfully ✅")
```

```python
# ===========================================
# Load Dataset
# ===========================================

df = pd.read_csv("../data/processed/zomato_processed.csv")

print("Dataset Loaded Successfully ✅")

print(df.shape)

df.head()
```

```python
# ===========================================
# Select Required Columns
# ===========================================

df = df[
    [
        "Review",
        "Rating",
        "Restaurant",
        "Cost",
        "Cuisines"
    ]
]

df.head()
```

```python
# ===========================================
# Clean Dataset
# ===========================================

# Remove 'Like'
df["Rating"] = df["Rating"].replace("Like", np.nan)

# Convert Rating to numeric
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

# Remove missing rows
df = df.dropna()

# Clean review text
df["Review"] = df["Review"].astype(str)

print(df.shape)

df.head()
```

```python
# ===========================================
# Convert Cost
# ===========================================

df["Cost"] = (
    df["Cost"]
      .astype(str)
      .str.replace(",", "", regex=False)
)

df["Cost"] = pd.to_numeric(
    df["Cost"],
    errors="coerce"
)

df["Cost"] = df["Cost"].fillna(df["Cost"].median())

print(df["Cost"].describe())
```

```python
# ===========================================
# Final Dataset Check
# ===========================================

print(df.info())

print("\nMissing Values:\n")

print(df.isnull().sum())
```

```python
# ===========================================
# Machine Learning Libraries
# ===========================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

print("ML Libraries Imported Successfully ✅")
```

```python
# ===========================================
# Features and Target
# ===========================================

X = df[
    [
        "Review",
        "Restaurant",
        "Cost",
        "Cuisines"
    ]
]

y = df["Rating"]

print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
# ===========================================
# Train Test Split
# ===========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data :", X_train.shape)

print("Testing Data :", X_test.shape)
```

```python
# ===========================================
# Preprocessing Pipeline
# ===========================================

preprocessor = ColumnTransformer(
    transformers=[
        ("review", TfidfVectorizer(max_features=5000), "Review"),
        ("restaurant", OneHotEncoder(handle_unknown="ignore"), ["Restaurant"]),
        ("cuisines", OneHotEncoder(handle_unknown="ignore"), ["Cuisines"]),
        ("cost", "passthrough", ["Cost"])
    ]
)

print("Preprocessor Created Successfully ✅")
```

```python
# ===========================================
# Import Machine Learning Models
# ===========================================

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Models Imported Successfully ✅")
```

```python
# ===========================================
# Model Pipelines
# ===========================================

lr_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        random_state=42
    ))
])

gb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingRegressor(
        random_state=42
    ))
])

print("Pipelines Created Successfully ✅")
```

```python
# ===========================================
# Train Models
# ===========================================

print("Training Linear Regression...")
lr_pipeline.fit(X_train, y_train)

print("Training Random Forest...")
rf_pipeline.fit(X_train, y_train)

print("Training Gradient Boosting...")
gb_pipeline.fit(X_train, y_train)

print("\nAll Models Trained Successfully ✅")
```

```python
# ===========================================
# Model Evaluation
# ===========================================

models = {
    "Linear Regression": lr_pipeline,
    "Random Forest": rf_pipeline,
    "Gradient Boosting": gb_pipeline
}

results = []

for name, model in models.items():

    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append([name, mae, rmse, r2])

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "RMSE",
        "R2 Score"
    ]
)

results_df.sort_values(
    by="R2 Score",
    ascending=False
)
```

```python
# ===========================================
# Best Model
# ===========================================

best_model = results_df.sort_values(
    by="R2 Score",
    ascending=False
).iloc[0]

print("🏆 Best Model")

print(best_model)
```

```python
# ===========================================
# Save Best Model
# ===========================================

import joblib

best_models = {
    "Linear Regression": lr_pipeline,
    "Random Forest": rf_pipeline,
    "Gradient Boosting": gb_pipeline
}

model_name = best_model["Model"]

joblib.dump(
    best_models[model_name],
    "../models/zomato_rating_model.pkl"
)

print("✅ Model Saved Successfully!")
```
