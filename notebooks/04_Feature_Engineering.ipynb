```python
# ===========================================
# Feature Engineering
# ===========================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

# Load processed dataset
df = pd.read_csv("../data/processed/zomato_processed.csv")

print("Dataset Loaded Successfully ✅")
```

```python
# ===========================================
# Clean Rating
# ===========================================

df["Rating"] = df["Rating"].replace("Like", np.nan)
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

print(df["Rating"].describe())
```

```python
# ===========================================
# Clean Cost
# ===========================================

df["Cost"] = (
    df["Cost"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df["Cost"] = pd.to_numeric(df["Cost"], errors="coerce")

print(df["Cost"].describe())
```

```python
# ===========================================
# Reviews Per Restaurant
# ===========================================

review_counts = (
    df.groupby("Restaurant")
      .size()
      .reset_index(name="Review_Count")
)

df = df.merge(review_counts, on="Restaurant", how="left")

df.head()
```

```python
# ===========================================
# Average Rating Per Restaurant
# ===========================================

avg_rating = (
    df.groupby("Restaurant")["Rating"]
      .mean()
      .reset_index(name="Average_Rating")
)

df = df.merge(avg_rating, on="Restaurant", how="left")

df.head()
```

```python
# ===========================================
# Review Length
# ===========================================

df["Review_Length"] = df["Review"].astype(str).apply(len)

df[["Review", "Review_Length"]].head()
```
