```python
# ===========================================
# Handle Missing Values
# ===========================================

# Metadata Dataset
metadata_df["Collections"] = metadata_df["Collections"].fillna("Unknown")
metadata_df["Timings"] = metadata_df["Timings"].fillna("Not Available")

# Reviews Dataset
reviews_df["Reviewer"] = reviews_df["Reviewer"].fillna("Anonymous")
reviews_df["Review"] = reviews_df["Review"].fillna("No Review")
reviews_df["Metadata"] = reviews_df["Metadata"].fillna("Unknown")
reviews_df["Time"] = reviews_df["Time"].fillna("Unknown")

# Rating (numeric column)
reviews_df["Rating"] = pd.to_numeric(reviews_df["Rating"], errors="coerce")
reviews_df["Rating"] = reviews_df["Rating"].fillna(reviews_df["Rating"].median())

# Verify that no missing values remain
print("Missing Values After Cleaning\n")

print("Metadata Dataset")
print(metadata_df.isnull().sum())

print("\n" + "="*50 + "\n")

print("Reviews Dataset")
print(reviews_df.isnull().sum())

print("\n✅ Missing values handled successfully!")
```

```python
# Standardize Restaurant Names

metadata_df["Name"] = (
    metadata_df["Name"]
    .str.strip()
    .str.lower()
)

reviews_df["Restaurant"] = (
    reviews_df["Restaurant"]
    .str.strip()
    .str.lower()
)

print("Restaurant Names Standardized ✅")
```

```python
# ===========================================
# Merge Metadata and Reviews
# ===========================================

merged_df = pd.merge(
    reviews_df,
    metadata_df,
    left_on="Restaurant",
    right_on="Name",
    how="inner"
)

print("Datasets Merged Successfully ✅")

print("\nMerged Dataset Shape :", merged_df.shape)
```
